```python
%%writefile app.py

# ==========================================================
# GLOBAL ECONOMIC INTELLIGENCE DASHBOARD
# Author: Aarav Singh
# ==========================================================

# ==========================
# IMPORT LIBRARIES
# ==========================

import streamlit as st
import pandas as pd
import plotly.express as px
import joblib

# ==========================
# PAGE CONFIGURATION
# ==========================

st.set_page_config(
    page_title="🌍 Global Economic Intelligence Dashboard",
    page_icon="🌍",
    layout="wide",
    initial_sidebar_state="expanded"
)

# ==========================
# LOAD DATASET
# ==========================

@st.cache_data
def load_data():
    df = pd.read_csv("global_economic_intelligence_dataset.csv")
    return df

df = load_data()

# ==========================
# LOAD MACHINE LEARNING MODEL
# ==========================

@st.cache_resource
def load_model():
    model = joblib.load("global_economic_model.pkl")
    return model

model = load_model()

# ==========================
# DASHBOARD TITLE
# ==========================

st.title("🌍 Global Economic Intelligence Dashboard")

st.markdown("""
### Interactive Analytics & Machine Learning Dashboard

This dashboard provides interactive visualizations of global economic
indicators and uses a trained Machine Learning model to predict
**Economic Growth Rate** based on economic indicators.
""")

st.markdown("---")
# ==========================================================
# SIDEBAR
# ==========================================================

st.sidebar.title("🌍 Dashboard Filters")

country = st.sidebar.selectbox(
    "Select Country",
    sorted(df["country"].unique())
)

filtered_df = df[df["country"] == country]

# Aggregate yearly values
yearly_df = (
    filtered_df
    .groupby("year", as_index=False)
    .mean(numeric_only=True)
)

st.sidebar.markdown("---")

st.sidebar.info(
    f"""
📌 **Selected Country**

{country}

📅 Years Available

{int(yearly_df['year'].min())} - {int(yearly_df['year'].max())}
"""
)

# ==========================================================
# KPI CARDS
# ==========================================================

st.subheader(f"📊 Economic Overview - {country}")

col1, col2, col3, col4 = st.columns(4)

with col1:

    latest_gdp = yearly_df.iloc[-1]["gdp_usd"]

    st.metric(
        label="GDP",
        value=f"${latest_gdp:,.0f}"
    )

with col2:

    latest_growth = yearly_df.iloc[-1]["economic_growth_rate"]

    st.metric(
        label="Growth Rate",
        value=f"{latest_growth:.2f}%"
    )

with col3:

    latest_inflation = yearly_df.iloc[-1]["inflation_rate"]

    st.metric(
        label="Inflation",
        value=f"{latest_inflation:.2f}%"
    )

with col4:

    latest_hdi = yearly_df.iloc[-1]["hdi"]

    st.metric(
        label="HDI",
        value=f"{latest_hdi:.3f}"
    )

st.markdown("---")
# ==========================================================
# GDP TREND
# ==========================================================

st.subheader("📈 GDP Trend")

fig = px.line(
    yearly_df,
    x="year",
    y="gdp_usd",
    markers=True,
    title=f"GDP Trend - {country}"
)

fig.update_layout(
    template="plotly_dark",
    height=450,
    xaxis_title="Year",
    yaxis_title="GDP (USD)"
)

st.plotly_chart(fig, use_container_width=True)

st.markdown("---")

# ==========================================================
# INFLATION & ECONOMIC GROWTH
# ==========================================================

col1, col2 = st.columns(2)

with col1:

    fig = px.bar(
        yearly_df,
        x="year",
        y="inflation_rate",
        color="inflation_rate",
        text_auto=".2f",
        title="📉 Inflation Rate"
    )

    fig.update_layout(
        template="plotly_dark",
        height=450,
        coloraxis_showscale=False
    )

    st.plotly_chart(fig, use_container_width=True)

with col2:

    fig = px.line(
        yearly_df,
        x="year",
        y="economic_growth_rate",
        markers=True,
        title="📈 Economic Growth Rate"
    )

    fig.update_layout(
        template="plotly_dark",
        height=450
    )

    st.plotly_chart(fig, use_container_width=True)

st.markdown("---")

# ==========================================================
# POPULATION & LIFE EXPECTANCY
# ==========================================================

col1, col2 = st.columns(2)

with col1:

    fig = px.area(
        yearly_df,
        x="year",
        y="population",
        title="👥 Population"
    )

    fig.update_layout(
        template="plotly_dark",
        height=450
    )

    st.plotly_chart(fig, use_container_width=True)

with col2:

    fig = px.line(
        yearly_df,
        x="year",
        y="life_expectancy",
        markers=True,
        title="❤️ Life Expectancy"
    )

    fig.update_layout(
        template="plotly_dark",
        height=450
    )

    st.plotly_chart(fig, use_container_width=True)

st.markdown("---")
# ==========================================================
# GDP PER CAPITA vs HDI
# ==========================================================

st.subheader("🌍 GDP Per Capita vs HDI")

fig = px.scatter(
    yearly_df,
    x="gdp_per_capita_usd",
    y="hdi",
    size="population",
    color="economic_growth_rate",
    hover_data=["year"],
    title=f"GDP Per Capita vs HDI - {country}"
)

fig.update_layout(
    template="plotly_dark",
    height=500
)

st.plotly_chart(fig, use_container_width=True)

st.markdown("---")

# ==========================================================
# CO2 EMISSIONS
# ==========================================================

st.subheader("🌱 CO₂ Emissions Per Capita")

fig = px.bar(
    yearly_df,
    x="year",
    y="co2_tonnes_per_capita",
    color="co2_tonnes_per_capita",
    text_auto=".2f",
    title=f"CO₂ Emissions - {country}"
)

fig.update_layout(
    template="plotly_dark",
    height=450,
    coloraxis_showscale=False
)

st.plotly_chart(fig, use_container_width=True)

st.markdown("---")

# ==========================================================
# GLOBAL GDP MAP
# ==========================================================

st.subheader("🗺️ Global GDP Distribution")

map_df = (
    df.groupby("country", as_index=False)["gdp_usd"]
      .mean()
)

country_mapping = {
    "USA": "United States",
    "UK": "United Kingdom"
}

map_df["country"] = map_df["country"].replace(country_mapping)

fig = px.choropleth(
    map_df,
    locations="country",
    locationmode="country names",
    color="gdp_usd",
    hover_name="country",
    color_continuous_scale="Viridis",
    title="Average GDP by Country"
)

fig.update_layout(
    template="plotly_dark",
    height=650,
    margin=dict(l=0, r=0, t=60, b=0)
)

st.plotly_chart(fig, use_container_width=True)

st.markdown("---")

# ==========================================================
# TOP 10 COUNTRIES BY GDP
# ==========================================================

st.subheader("🏆 Top 10 Countries by Average GDP")

top10 = (
    df.groupby("country", as_index=False)["gdp_usd"]
      .mean()
      .sort_values("gdp_usd", ascending=False)
      .head(10)
)

fig = px.bar(
    top10,
    x="country",
    y="gdp_usd",
    color="gdp_usd",
    text_auto=".2s",
    title="Top 10 Countries by Average GDP"
)

fig.update_layout(
    template="plotly_dark",
    height=500,
    showlegend=False,
    xaxis_title="Country",
    yaxis_title="Average GDP (USD)"
)

st.plotly_chart(fig, use_container_width=True)

st.markdown("---")
# ==========================================================
# AI ECONOMIC GROWTH PREDICTOR
# ==========================================================

st.markdown("---")

st.header("🤖 AI Economic Growth Predictor")

st.write(
    "Enter the economic indicators below to predict the expected Economic Growth Rate using the trained Machine Learning model."
)

col1, col2 = st.columns(2)

with col1:

    pred_country = st.selectbox(
        "Country",
        sorted(df["country"].unique()),
        key="prediction_country"
    )

    pred_year = st.number_input(
        "Year",
        min_value=2000,
        max_value=2050,
        value=2025
    )

    pred_population = st.number_input(
        "Population",
        min_value=100000,
        value=50000000,
        step=100000
    )

    pred_gdp = st.number_input(
        "GDP (USD)",
        min_value=1000000.0,
        value=1000000000000.0,
        step=1000000.0,
        format="%.0f"
    )

    pred_gdp_per_capita = st.number_input(
        "GDP Per Capita (USD)",
        min_value=100.0,
        value=25000.0
    )

with col2:

    pred_inflation = st.number_input(
        "Inflation Rate (%)",
        value=4.0
    )

    pred_unemployment = st.number_input(
        "Unemployment Rate (%)",
        value=6.0
    )

    pred_life = st.number_input(
        "Life Expectancy",
        value=74.0
    )

    pred_hdi = st.number_input(
        "HDI",
        min_value=0.0,
        max_value=1.0,
        value=0.800,
        format="%.3f"
    )

    pred_co2 = st.number_input(
        "CO₂ Tonnes Per Capita",
        value=5.0
    )

st.markdown("")

if st.button("🚀 Predict Economic Growth"):

    input_df = pd.DataFrame({
        "country":[pred_country],
        "year":[pred_year],
        "population":[pred_population],
        "gdp_usd":[pred_gdp],
        "gdp_per_capita_usd":[pred_gdp_per_capita],
        "inflation_rate":[pred_inflation],
        "unemployment_rate":[pred_unemployment],
        "life_expectancy":[pred_life],
        "hdi":[pred_hdi],
        "co2_tonnes_per_capita":[pred_co2]
    })

    prediction = model.predict(input_df)[0]

    st.success(
        f"### 🌍 Predicted Economic Growth Rate: **{prediction:.2f}%**"
    )

    if prediction >= 6:
        st.info("📈 Strong projected economic growth.")

    elif prediction >= 3:
        st.info("📊 Moderate projected economic growth.")

    elif prediction >= 0:
        st.warning("⚠️ Slow projected economic growth.")

    else:
        st.error("📉 Negative economic growth (economic contraction) predicted.")
# ==========================================================
# DOWNLOAD FILTERED DATASET
# ==========================================================

st.markdown("---")

st.header("📥 Download Dataset")

csv = yearly_df.to_csv(index=False).encode("utf-8")

st.download_button(
    label="📥 Download Current Country Dataset",
    data=csv,
    file_name=f"{country}_economic_data.csv",
    mime="text/csv"
)

# ==========================================================
# DATASET PREVIEW
# ==========================================================

st.markdown("---")

st.header("📋 Dataset Preview")

st.dataframe(
    yearly_df,
    use_container_width=True,
    hide_index=True
)

# ==========================================================
# SUMMARY STATISTICS
# ==========================================================

st.markdown("---")

st.header("📊 Summary Statistics")

summary = yearly_df.describe().T

summary = summary[[
    "mean",
    "std",
    "min",
    "25%",
    "50%",
    "75%",
    "max"
]]

st.dataframe(
    summary,
    use_container_width=True
)

# ==========================================================
# COUNTRY INSIGHTS
# ==========================================================

st.markdown("---")

st.header("📌 Country Insights")

c1, c2 = st.columns(2)

with c1:

    st.metric(
        "Highest GDP",
        f"${yearly_df['gdp_usd'].max():,.0f}"
    )

    st.metric(
        "Highest GDP Per Capita",
        f"${yearly_df['gdp_per_capita_usd'].max():,.0f}"
    )

    st.metric(
        "Highest HDI",
        f"{yearly_df['hdi'].max():.3f}"
    )

with c2:

    st.metric(
        "Lowest Inflation",
        f"{yearly_df['inflation_rate'].min():.2f}%"
    )

    st.metric(
        "Highest Life Expectancy",
        f"{yearly_df['life_expectancy'].max():.2f} Years"
    )

    st.metric(
        "Highest Growth Rate",
        f"{yearly_df['economic_growth_rate'].max():.2f}%"
    )

# ==========================================================
# PROJECT INFORMATION
# ==========================================================

st.markdown("---")

with st.expander("ℹ️ About This Project"):

    st.markdown("""
### 🌍 Global Economic Intelligence Dashboard

This project combines **Data Analytics**, **Machine Learning**, and
**Interactive Data Visualization**.

### Features

- 📊 Interactive Economic Dashboard
- 🌍 World GDP Map
- 📈 GDP & Population Trends
- 📉 Inflation & Growth Analysis
- 🤖 AI-Based Economic Growth Prediction
- 📥 Download Filtered Dataset
- 📋 Summary Statistics

### Machine Learning Model

- Random Forest Regressor
- One-Hot Encoding
- Scikit-Learn Pipeline

### Technologies Used

- Python
- Streamlit
- Pandas
- Plotly
- Scikit-Learn
- Joblib
""")

# ==========================================================
# FOOTER
# ==========================================================

st.markdown("---")

st.markdown(
"""
<div style='text-align:center; color:gray;'>

<h2>🌍 Global Economic Intelligence Dashboard</h2>

<p>
Developed using Python • Streamlit • Plotly • Scikit-Learn
</p>

<p>
Interactive Data Analytics & Machine Learning Dashboard
</p>

<hr>

<p>
© 2026 Aarav Singh
</p>

</div>
""",
unsafe_allow_html=True
)
```
